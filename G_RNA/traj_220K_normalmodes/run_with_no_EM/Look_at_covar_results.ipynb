```python
# Specify coordinates we want
O6term = u.select_atoms('name O6')[0]
C6term = u.select_atoms('name C6')[0]
N2term = u.select_atoms('name N2')[0]
H21term = u.select_atoms('name H21')[0]
H22term = u.select_atoms('name H22')[0]

# Extract some coordinates (and time)
O6_position = []
C6_position = []
N2_position = []
H21_position = []
H22_position = []
time = []
for ts in u.trajectory:     # iterate through all frames
    O6_position.append(O6term.position)
    C6_position.append(C6term.position)
    N2_position.append(N2term.position)
    H21_position.append(H21term.position)
    H22_position.append(H22term.position)
    time.append(u.trajectory.time)
```

### Analyzing spectra of filtered modes

```python
# First, some common values
dt = time[1]-time[0]  # Sampling interval
N = len(time)
c = 3.0e10 # speed of light, m/s
f = np.fft.fftfreq(N, dt)  # Frequency vector
f_Hz = f * 1e12
nubar = inverseps2wavenum(f)
vertical_amp = 1e5
```

```python
# Get the carbonyl bond length
vector_difference = np.array(C6_position)-np.array(O6_position)
carbonyl_distance = np.sqrt(np.sum(vector_difference**2,axis=1))
carbonyl_distance_avg = np.mean(carbonyl_distance)
carbonyl_deviation = carbonyl_distance-carbonyl_distance_avg
carbonyl_rms_distance = np.sqrt(np.mean((carbonyl_deviation)**2))
percent_deviation = carbonyl_rms_distance/carbonyl_distance_avg*100
print('avg carbonyl_distance', carbonyl_distance_avg)
print('carbonyl_rms_distance', carbonyl_rms_distance)

# Plot it
plt.figure()
plt.plot(time,carbonyl_deviation)
plt.title(CWD+', C=O'+' %deviation = '+str(percent_deviation))
plt.grid(True)
plt.xlabel('time [ps]')
plt.ylabel('deviation [A]')

# Compute FFT and power spectral density
Y = np.fft.fft(carbonyl_distance) / N  # Normalizes
Pyy = 2 * np.abs(Y)**2

# Plot it
fix,ax = plt.subplots(layout='constrained')
ixstart = 2000 # Skips DC and other low-frequency components
plt.plot(nubar[ixstart:N//2], Pyy[ixstart:N//2]*vertical_amp)  # Plot only positive frequencies
plt.ylabel('Power Spectral Density (x10^5)')
plt.title('Power Spectrum of '+xtcfile+' '+CWD)
xlim_inverseps = np.array([40,60])
plt.xlabel('Frequency [1/cm]')
xlim_wavenum = np.array([1300,1950])
plt.xlim(xlim_wavenum)
plt.grid()

# This adds tickmarks on the top
secax = ax.secondary_xaxis('top', functions=(wavenum2inverseps,inverseps2wavenum))
secax.set_xlabel('wavenum [1/ps]')
```

```python
# Get the amine angle
N2_H21_vectors = np.array(N2_position)-np.array(H21_position)
N2_H22_vectors = np.array(N2_position)-np.array(H22_position)
scissor_angle = np.zeros(N)

for i in range(N):
    dot_product = np.dot(N2_H21_vectors[i,:],N2_H22_vectors[i,:])
    magnitude_N2_H21 = np.sqrt(np.dot(N2_H21_vectors[i,:],N2_H21_vectors[i,:]))
    magnitude_N2_H22 = np.sqrt(np.dot(N2_H22_vectors[i,:],N2_H22_vectors[i,:]))
    scissor_angle[i] = np.arccos(dot_product/(magnitude_N2_H21*magnitude_N2_H22))*180/np.pi
scissor_angle_mean = np.mean(scissor_angle)
scissor_angle_deviation = scissor_angle - scissor_angle_mean
scissor_angle_rms_difference = np.sqrt(np.mean((scissor_angle_deviation)**2))
percent_deviation = scissor_angle_rms_difference/scissor_angle_mean*100
print('avg scissor angle', scissor_angle_mean)
print('scissor_rms_difference', scissor_angle_rms_difference)

# Plot it
plt.figure()
plt.plot(time,scissor_angle_deviation)
plt.grid(True)
plt.xlabel('time [ps]')
plt.ylabel('scissor angle deviation [deg]')
plt.title(CWD+', amine'+' %deviation = '+str(percent_deviation))

# Compute FFT and power spectral density
Y = np.fft.fft(scissor_angle_deviation) / N  # Normalizes
Pyy = 2 * np.abs(Y)**2

# Plot it
fix,ax = plt.subplots(layout='constrained')
ixstart = 0 # Skips DC and other low-frequency components
plt.plot(nubar[ixstart:N//2], Pyy[ixstart:N//2]*vertical_amp)  # Plot only positive frequencies
plt.ylabel('Power Spectral Density (x10^5)')
plt.title('Power Spectrum of '+xtcfile+' '+CWD)
xlim_wavenum = np.array([1400,1900])
plt.xlabel('wavenum [1/cm]')
plt.xlim(xlim_wavenum)
plt.grid()

# This adds tickmarks on the top
secax = ax.secondary_xaxis('top', functions=(wavenum2inverseps,inverseps2wavenum))
secax.set_xlabel('frequency [1/ps]')
```
