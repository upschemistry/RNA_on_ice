```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import uniform_filter1d
from matplotlib import colors
print(colors.CSS4_COLORS.keys())
```

```python
OP_220K = np.loadtxt("traj_220K_100ns/OP_220.txt")
OP_245K = np.loadtxt("traj_245K_100ns/OP.txt")
```

```python
print(np.shape(OP_220K))
time_220K = OP_220K[0,:]
O1P_220K = OP_220K[1,:]
O2P_220K = OP_220K[2,:]
P_220K = OP_220K[3,:]
time_245K= OP_245K[0,:]
O1P_245K = OP_245K[1,:]
O2P_245K = OP_245K[2,:]
P_245K = OP_245K[3,:]
smoothsize = 7
O1P_220K = uniform_filter1d(O1P_220K,size=smoothsize)
O2P_220K = uniform_filter1d(O2P_220K,size=smoothsize)
P_220K = uniform_filter1d(P_220K,size=smoothsize)
O1P_245K = uniform_filter1d(O1P_245K,size=smoothsize)
O2P_245K = uniform_filter1d(O2P_245K,size=smoothsize)
P_245K = uniform_filter1d(P_245K,size=smoothsize)
```

```python
plt.figure()
plt.plot(time_220K,O1P_220K, "darkcyan",label='O1P 220K')
plt.plot(time_220K,O2P_220K, "blue",label='O2P 220K')
plt.plot(time_220K,P_220K, "darkblue",label='P 220K')
plt.plot(time_245K,O1P_245K, "sienna",label='O1P 245K')
plt.plot(time_245K,O2P_245K, "red",label='O2P 245K')
plt.plot(time_245K,P_245K, "darkorange",label='P 245K')
plt.xlabel('time (ns)')
plt.ylabel('distance (nm)')
plt.legend()
plt.grid(True)
```
