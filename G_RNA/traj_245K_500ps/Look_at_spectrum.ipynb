```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# Use np.loadtxt to load in vac.xvg
data_245K = np.loadtxt('spectrum.xvg',comments=['@','#','&'])
data_220K = np.loadtxt('../traj_220K_500ps/spectrum.xvg',comments=['@','#','&'])
data_270K = np.loadtxt('../traj_270K_500ps/spectrum.xvg',comments=['@','#','&'])
```

```python
# Use plt.plot to plot 2nd column on the y-axis, 1st column on the x-axis
nu_wavenum = data_245K[:,0]
print('spectral resolution =',nu_wavenum[1]-nu_wavenum[0])
plt.figure()
plt.plot(nu_wavenum,data_220K[:,1],color = 'blue', label='G 220 K', alpha=.6)
plt.plot(nu_wavenum,data_245K[:,1],color = 'red', label='G 245 K', alpha=.6)
#plt.plot(nu_wavenum,data_270K[:,1],color = 'red', label='270 K', alpha=.6)
plt.xlabel('wavenumber (cm-1)')
plt.ylabel('power (arb. units)')
plt.grid(True)
plt.xlim([1600,1900])
plt.ylim([0,200])
plt.legend()
```
