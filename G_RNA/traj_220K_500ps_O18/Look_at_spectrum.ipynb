```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib notebook
```

```python
# Use np.loadtxt to load in vac.xvg
data_220K_O18 = np.loadtxt('spectrum.xvg',comments=['@','#','&'])
data_220K = np.loadtxt('../traj_220K_500ps/spectrum.xvg',comments=['@','#','&'])
data_220K_O99 = np.loadtxt('../traj_220K_500ps_O99/spectrum.xvg',comments=['@','#','&'])
```

```python
# Use plt.plot to plot 2nd column on the y-axis, 1st column on the x-axis
nu_wavenum = data_220K[:,0]
print('spectral resolution =',nu_wavenum[1]-nu_wavenum[0])
plt.figure()
plt.plot(nu_wavenum,data_220K[:,1],color = 'blue', label='220 K')
plt.plot(nu_wavenum,data_220K_O18[:,1],color = 'k', label='220 K, w/Oxygen 18 on carbonyl', alpha=.4)
plt.plot(nu_wavenum,data_220K_O99[:,1],color = 'red', label='220 K, w/Oxygen 99 on carbonyl', alpha=.4)
plt.xlabel('wavenumber (cm-1)')
plt.ylabel('power (arb. units)')
plt.grid(True)
plt.xlim([1550,1950])
plt.legend()
```
