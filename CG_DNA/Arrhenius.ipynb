```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib notebook
```

```python
# Units = K
T = np.array([230, 236])

# Units 1/K
beta = 1/T

# Units = cm^2/s, converting to SI, m^2/s (although the units don't matter for calculating deltaH)
D = np.array([2.045e-05, 0.001677])*1e-5 
D *= (1/100)**2

# Graphics
plt.figure()
plt.plot(beta*1e3,np.log(D),'o-')
plt.xlabel('1/T x 1000')
plt.ylabel('ln(D)')
```

```python
# Universal gas constant, units = kJ/mol/K
R = 8.314/1e3

# Arrhenius slope
delta_beta = beta[1]-beta[0]
delta_lnD = np.log(D[1]/D[0])

# Enthalpy
deltaH = -R * delta_lnD/delta_beta
print(deltaH)
```
