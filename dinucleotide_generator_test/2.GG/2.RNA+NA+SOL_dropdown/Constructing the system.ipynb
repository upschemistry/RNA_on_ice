```python
# Loads just a .gro single-frame file and checks if there are positions and velocities
grofilename =   'SOL_annealed.gro'; print(grofilename)
uframe1 = mda.Universe(grofilename)
print(uframe1.trajectory.ts.has_positions)
print(uframe1.trajectory.ts.has_velocities)

# Show the whole system
TOTview = nv.show_mdanalysis(uframe1)
TOTview.clear()
TOTview.add_representation('ball+stick',selection='SYSTEM')
TOTview.camera='orthographic'
TOTview.center()
TOTview.control.rotate([0,1,0,0])
TOTview.add_unitcell()
TOTview
```

```python
# Loads just a .gro single-frame file and checks if there are positions and velocities
grofilename =   '../1.RNA/confout_shifted.gro'; print(grofilename)
uframe2 = mda.Universe(grofilename)
print(uframe2.trajectory.ts.has_positions)
print(uframe2.trajectory.ts.has_velocities)

# Show the whole system
TOTview = nv.show_mdanalysis(uframe2)
TOTview.clear()
TOTview.add_representation('ball+stick',selection='RNA')
TOTview.add_representation('axes',selection='RNA')
TOTview.camera='orthographic'
TOTview.center()
TOTview.control.rotate([0,1,0,0])
TOTview
```

```python
grofilename =   'NA.gro'; print(grofilename)
uframe3 = mda.Universe(grofilename)
print(uframe3.trajectory.ts.has_positions)
print(uframe3.trajectory.ts.has_velocities)
```

```python
# Make the entire .gro 
SOL = uframe1.select_atoms('segid SYSTEM')
RNA = uframe2.select_atoms('segid SYSTEM')
NA = uframe3.select_atoms('segid SYSTEM')
uframe4 = mda.Merge(RNA,NA,SOL)

# Writing
writing = True
if writing:
    grofilename = "System1.gro"
    TOT = uframe4.select_atoms('segid SYSTEM')
    uframe4.dimensions = uframe1.dimensions
    TOT.write(grofilename)

# Show the whole system
TOTview = nv.show_mdanalysis(uframe4)
TOTview.clear()
TOTview.add_representation('licorice',selection='SOL')
TOTview.add_unitcell()
TOTview.add_representation('ball+stick',selection='RNA')
TOTview.add_representation('ball+stick',selection='NA',radius=0.5)
TOTview.add_representation('axes',selection='RNA')
TOTview.camera='orthographic'
TOTview.center()
TOTview.control.rotate([0,1,0,0])
TOTview.add_unitcell()
TOTview
```

```python
# Checking
grofilename = "System1.gro"
uframe5 = mda.Universe(grofilename)
print(uframe5.trajectory.ts.has_positions)
print(uframe5.trajectory.ts.has_velocities)

# Show the whole system
TOTview = nv.show_mdanalysis(uframe5)
TOTview.clear()
TOTview.add_representation('licorice',selection='SOL')
TOTview.add_representation('ball+stick',selection='RNA')
TOTview.add_representation('ball+stick',selection='NA',radius=0.5)
TOTview.camera='orthographic'
TOTview.center()
TOTview.add_unitcell()
TOTview.control.rotate([0,1,0,0])

TOTview
```
